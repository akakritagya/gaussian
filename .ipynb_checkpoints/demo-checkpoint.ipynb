# Gaussian Elimination — Demo

This notebook walks you through a variety of test scenarios for the `Gaussian` class.


#### Setup

Load packages


```python
import numpy as np
from gaussian import Gaussian
```

---

#### Case 1

Simple 2-Variable System


```python
equations_2var = ["x + 2y = 5", "3x - y = 4"]

ge1 = Gaussian(equations_2var)
ge1.solve()
ge1.echelon_M()
ge1.system()
ge1.show_solution()
print(repr(ge1))
print(ge1)
```

#### Case 2

System with 3 variables


```python
equations_3var = ["2x + y - z = 8", "-3x - y + 2z = -11", "-2x + y + 2z = -3"]

ge = Gaussian(equations_3var)
ge.solve()
ge.system()
ge.show_solution()
print(repr(ge))
print(ge)
```

#### Case 3

System with large number of variables


```python
equations_many_vars = [
    "10a + 2b - c + d + e - f + g + h - i + j + k = 15",
    "a + 9b + 2c - d + e + f - g + h + i - j + k = 14",
    "2a - b + 10c + d - e + f + g - h + i + j - k = 13",
    "a + b - 2c + 11d + e + f - g + h - i + j + k = 14",
    "-a + b + c + 2d + 12e - f + g + h + i - j + k = 17",
    "a - b + c - d + e + 10f + g - h + i + j + 2k = 15",
    "a + b - c + d - e + f + 9g + h + i - j + k = 13",
    "2a + b + c - d + e - f + g + 10h + i + j - k = 15",
    "a - b + 2c + d + e + f - g + h + 11i - j + k = 16",
    "a + b - c + 2d - e + f + g + h + i + 10j + k = 17",
    "a + b + c - d + e - f + 2g - h + i + j + 12k = 17",
]

ge = Gaussian(equations_many_vars)
ge.solve()
ge.system()
ge.show_solution()
print(repr(ge))
print(ge)
```

#### Case 4

Homogenous system


```python
equations = ["2x + y - z = 0", "x - 3y + 2z = 0", "3x + 2y + z = 0"]

ge = Gaussian(equations)
ge.solve()
ge.system()
ge.show_solution()
print(repr(ge))
print(ge)
```

#### Case 5

System with unique solution (complete system)


```python
equations = ["2x + y - z = 8", "x + 3y + 2z = 17", "3x - y + z = 6"]

ge = Gaussian(equations)
ge.solve()
ge.system()
ge.show_solution()
print(repr(ge))
print(ge)
```

#### Case 6

System with infinitely many solutions (redundant system)


```python
equations = [
    "x + 2y + 3z = 6",
    "2x + 4y + 6z = 12",  # 2x row 1 — linearly dependent
    "3x + 6y + 9z = 18",  # 3x row 1 — linearly dependent
]

ge = Gaussian(equations)
ge.solve()
ge.echelon_M()
ge.system()
ge.show_solution()
print(repr(ge))
print(ge)
```

#### Case 7

System with no solution (inconsistent system)


```python
equations = [
    "x + 2y + 3z = 6",
    "2x + 4y + 6z = 12",  # 2x row 1 — linearly dependent
    "3x + 6y + 9z = 25",  # same LHS pattern but contradictory RHS
]

ge = Gaussian(equations)
ge.solve()
ge.system()
ge.show_solution()
print(repr(ge))
print(ge)
```

#### Case 8

System requiring row swaps (partial pivoting)


```python
equations = [
    "2b + 3c - d + e = 10",  # no 'a' term → zero pivot in column 0, row swap required
    "4a - 2b + c + 3d - e = 12",
    "3a + b - 2c + d + 2e = 11",
    "2a + 3b + c - 3d + e = 8",
    "a - b + 4c + 2d - 2e = 6",
]

ge = Gaussian(equations)
ge.solve()
ge.augmented_M()
ge.system()
ge.show_solution()
print(repr(ge))
print(ge)
```

#### Case 9

Compare `Gaussian.solve()` with `np.linalg.solve()`


```python
equations = [
    "2x + y - 3z + 2w = 10",
    "x - 3y + 2z - w = -5",
    "3x + 2y + z + 4w = 20",
    "x + y - z + 3w = 8",
]

ge = Gaussian(equations)
ge.solve()
ge.system()
ge.show_solution()
print(repr(ge))
print(ge)

# numpy cross-check
A = np.array([[2, 2, 1, -3], [-1, 1, -3, 2], [4, 3, 2, 1], [3, 1, 1, -1]], dtype=float)
b = np.array([10, -5, 20, 8], dtype=float)

np_solution = np.linalg.solve(A, b)
np_solution_dict = dict(
    zip(["w", "x", "y", "z"], np_solution.round(6))
)  # sorted alphabetically to match Gaussian

print("\n  Gaussian solution :", ge.solution)
print("  numpy solution    :", np_solution_dict)

if ge.solution:
    match = all(
        np.isclose(ge.solution[k], np_solution_dict[k], atol=1e-8) for k in ge.solution
    )
    print(f"\n  Solutions match   : {match}")
```

---

#### Summary of Test Cases

| Case | Description                   | Variables     | Expected Outcome      |
| ---- | ----------------------------- | ------------- | --------------------- |
| 1    | Simple 2-variable system      | x, y          | Unique solution       |
| 2    | 3-variable system             | x, y, z       | Unique solution       |
| 3    | Large 11-variable system      | a-k           | All variables = 1.0   |
| 4    | Trivial solution              | x, y, z       | All variables = 0.0   |
| 5    | Unique solution               | x, y, z       | Single solution point |
| 6    | Infinitely many solutions     | x, y, z       | Singular Matrix, None |
| 7    | No solution (inconsistent)    | x, y, z       | Singular Matrix, None |
| 8    | Requires row swaps            | a, b, c, d, e | Unique solution       |
| 9    | Gaussian vs np.linalg.solve() | w, x, y, z    | Solutions match       |
